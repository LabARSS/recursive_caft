```python
from pathlib import Path

import pandas as pd

SRC_PATH = Path("../../data/out/distillation/mmlu_distilled_deepseek_v4_flash_extend_w_large.parquet")
OUT_PATH = Path("../../data/out/distillation/mmlu_distilled_deepseek_v4_flash_extend_w_large_incorrect_only.parquet")

CORRECT_COL = "distill_ans_correct"

df = pd.read_parquet(SRC_PATH)
print("Source shape:", df.shape)
print("Correctness:\n", df[CORRECT_COL].value_counts(dropna=False))
```

```python
incorrect_df = df[df[CORRECT_COL] == False].reset_index(drop=True)
print("Incorrect shape:", incorrect_df.shape)
```

```python
OUT_PATH.parent.mkdir(parents=True, exist_ok=True)
incorrect_df.to_parquet(OUT_PATH, index=False)
print(f"Wrote {len(incorrect_df)} rows to {OUT_PATH}")
```
