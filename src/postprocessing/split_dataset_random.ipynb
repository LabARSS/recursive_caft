```python
from pathlib import Path

import numpy as np
import pandas as pd

from core.utils.seed import set_seed

set_seed()
np.random.seed(42)

TEST_ALLOCATION = 0.2
SPLIT_OUT_PATH = Path("../../data/out/splits/random/mmlu/")

df = pd.read_parquet("../../data/source/mmlu_pro_stem.parquet")
print(f"Total samples: {len(df)}")
df.head()
```

```python
from core.utils.splitter import split_chunk_into_train_test

train_df, test_df = split_chunk_into_train_test(df, TEST_ALLOCATION)

print(f"Train len: {len(train_df)}")
print(f"Test len: {len(test_df)}")
print(f"Test ratio: {len(test_df) / (len(train_df) + len(test_df)):.4f}")
```

```python
SPLIT_OUT_PATH.mkdir(parents=True, exist_ok=True)

train_df.to_parquet(str(SPLIT_OUT_PATH / "train.parquet"), index=False)
test_df.to_parquet(str(SPLIT_OUT_PATH / "test.parquet"), index=False)

print(f"Saved to {SPLIT_OUT_PATH.resolve()}")
```
